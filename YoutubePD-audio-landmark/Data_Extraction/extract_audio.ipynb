```python
def get_sec(tstr):
    if(isinstance(tstr, str)):
        v = tstr.split(":")
        m, s = v[0], v[1]
    else:
        m, s = tstr.hour, tstr.minute
    if(int(m) < 10):
        m = '0'+str(m)
    if(int(s) < 10):
        s = '0'+str(s)
    return "00:"+str(m)+":"+str(s)

train_list = []
test_list = []
for l in open('../../YouTubePD-data/train.csv'):
    label = int(l.split(' ')[-1])
    id = int(l.split(' ')[0].split('/')[-1].split('_')[0][len('video'):])
    train_list.append(id)

for l in open('../../YouTubePD-data/test.csv'):
    label = int(l.split(' ')[-1])
    id = int(l.split(' ')[0].split('/')[-1].split('_')[0][len('video'):])
    test_list.append(id)


import os
import pandas as pd
import subprocess

df = pd.read_excel('../../YouTubePD-data/data_sheets/data_sheet.xlsx')
for i in range(df.shape[0]):
    datapt = df.iloc[i]
    ids = datapt['link'].split('=')[-1].strip()
    if i in train_list:
        split = 'train'
    elif i in test_list:
        split = 'test'
    else:
        continue
    src = 'standardized_split_video/' + split + "/" + str(i) + datapt['parkinson y/n'].strip()+".mp4"
    dest = 'standardized_split_audio/' + split + "/" + str(i) + datapt['parkinson y/n'].strip()+".wav"
    cmd = "C:/ffmpeg/bin/ffmpeg.exe -i "+src+" -ab 160k -ac 2 -ar 44100 -vn "+dest
    print(cmd)

shift = df.shape[0]

df = pd.read_excel('../../YouTubePD-data/data_sheets/NegSamples.xlsx')
for i in range(df.shape[0]):
    datapt = df.iloc[i]
    ids = datapt['link'].split('=')[-1].strip()
    if i+shift in train_list:
        split = 'train'
    elif i+shift in test_list:
        split = 'test'
    else:
        continue
    src = 'standardized_split_video/' + split + "/" + str(i+shift) + 'n'.strip()+".mp4"
    dest = 'standardized_split_audio/' + split + "/" + str(i+shift) + 'n'.strip()+".wav"
    cmd = "C:/ffmpeg/bin/ffmpeg.exe -i "+src+" -ab 160k -ac 2 -ar 44100 -vn "+dest
    print(cmd)
```
