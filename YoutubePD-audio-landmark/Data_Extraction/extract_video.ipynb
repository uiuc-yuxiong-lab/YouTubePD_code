```python
def get_sec(tstr):
    if(isinstance(tstr, str)):
        v = tstr.split(":")
        m, s = v[0], v[1]
    else:
        m, s = tstr.hour, tstr.minute
    if(int(m) < 10):
        m = '0'+str(int(m))
    if(int(s) < 10):
        s = '0'+str(int(s))
    return "00:"+str(m)+":"+str(s)

import os
import pandas as pd
import subprocess


train_list = []
test_list = []
for l in open('../../YouTubePD-data/train.csv'):
    label = int(l.split(' ')[-1])
    id = int(l.split(' ')[0].split('/')[-1].split('_')[0][len('video'):])
    train_list.append(id)

for l in open('../../YouTubePD-data/test.csv'):
    label = int(l.split(' ')[-1])
    id = int(l.split(' ')[0].split('/')[-1].split('_')[0][len('video'):])
    test_list.append(id)



df = pd.read_excel('../../YouTubePD-data/data_sheets/data_sheet.xlsx')
for i in range(df.shape[0]):
    datapt = df.iloc[i]
    start_time = get_sec(datapt['start'])
    end_time = get_sec(datapt['end'])

    link = datapt['link'].strip()
    ids = datapt['link'].split('=')[-1].strip()
    if i in train_list:
        split = 'train'
    elif i in test_list:
        split = 'test'
    else:
        continue
    destination = 'standardized_split_video\\' + split + "\\" + str(i) + datapt['parkinson y/n'].strip()+".mp4"

    
    cmd = "C:\\ffmpeg\\bin\\ffmpeg.exe -ss "+start_time+" -to "+end_time+" -i \"$(youtube-dl --rm-cache-dir -f best --get-url '"+link+"')\" -c:v copy -c:a copy "+destination
    print(cmd)

shift = df.shape[0]

df = pd.read_excel('../../YouTubePD-data/data_sheets/NegSamples.xlsx')
for i in range(df.shape[0]):
    datapt = df.iloc[i]
    start_time = get_sec(datapt['start'])
    end_time = get_sec(datapt['end'])

    link = datapt['link'].strip()
    ids = datapt['link'].split('=')[-1].strip()
    if i+shift in train_list:
        split = 'train'
    elif i+shift in test_list:
        split = 'test'
    else:
        continue
    destination = 'standardized_split_video\\' + split + "\\" + str(i+shift) + 'n'.strip()+".mp4"

    
    cmd = "C:\\ffmpeg\\bin\\ffmpeg.exe -ss "+start_time+" -to "+end_time+" -i \"$(youtube-dl --rm-cache-dir -f best --get-url '"+link+"')\" -c:v copy -c:a copy "+destination
    print(cmd)
```
